```python
import numpy as np

print("NumPy version:", np.__version__)
print("Day 4 environment is ready!")
```

```python
tokens = ["AI", "learns", "patterns"]

E = np.array([
    [1, 0, 1, 0],  # AI
    [0, 1, 0, 1],  # learns
    [1, 1, 0, 0]   # patterns
], dtype=float)

print("Tokens:", tokens)
print("Embedding matrix:")
print(E)

print("Matrix shape:", E.shape)
```

```python
Q = E
K = E
V = E

d_k = K.shape[1]

scores = (Q @ K.T) / np.sqrt(d_k)

print("Attention scores:")
print(np.round(scores, 2))
```

```python
def softmax(x, axis=-1):
    x = x - np.max(x, axis=axis, keepdims=True)

    exp_x = np.exp(x)

    return exp_x / np.sum(
        exp_x,
        axis=axis,
        keepdims=True
    )


weights = softmax(scores)

print("Attention weights:")
print(np.round(weights, 2))

print("Row sums:")
print(np.round(weights.sum(axis=1), 2))
```

```python
context_vectors = weights @ V

print("Contextualized token representations:")
print(np.round(context_vectors, 2))

print("Output shape:", context_vectors.shape)
```

```python
import math

def is_prime(n):
    if n <= 1:
        return False

    if n == 2:
        return True

    if n % 2 == 0:
        return False

    for i in range(3, int(math.sqrt(n)) + 1, 2):
        if n % i == 0:
            return False

    return True


print(is_prime(7))   # True
print(is_prime(10))  # False
print(is_prime(29))  # True
print(is_prime(49))  # False
```
